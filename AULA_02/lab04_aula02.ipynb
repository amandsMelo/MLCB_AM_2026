```python
# ============================================================
# LAB 04 - AULA 02 (MLCB): DESAFIO NLU PARA AGÊNCIA DE VIAGENS
# ============================================================

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

# 1. Criando o dataset
dados_viagens = {
    'mensagem': [
        'Quero comprar uma passagem para São Paulo','Preciso de uma passagem para o Rio de Janeiro',
        'Gostaria de reservar um voo para Salvador','Quero comprar uma passagem de avião',
        'Quero cancelar minha reserva','Preciso cancelar minha passagem',
        'Como faço para cancelar meu voo?','Quero desistir da minha reserva','Quero falar com um atendente',
        'Preciso de ajuda de um atendente','Gostaria de falar com uma pessoa','Pode me passar para o atendimento?',
        'Quero falar com um funcionário da agência','Preciso falar com um atendente sobre minha viagem',
        'Tem alguém que possa me atender?'

    ],
    'intencao': [
        'comprar_passagem','comprar_passagem','comprar_passagem',
        'comprar_passagem','cancelar_reserva','cancelar_reserva',
        'cancelar_reserva','cancelar_reserva','falar_atendente',
        'falar_atendente','falar_atendente','falar_atendente',
        'falar_atendente','falar_atendente','falar_atendente'

    ]
}

df4 = pd.DataFrame(dados_viagens)

# 2. Separando X e y
X = df4['mensagem']
y = df4['intencao']

# 3. Dividindo os dados em treino e teste
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# 4. Vetorização com TF-IDF
vectorizer = TfidfVectorizer()

X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# 5. Criando e treinando o modelo
modelo = LogisticRegression()

modelo.fit(X_train_vec, y_train)

# 6. Testando com frases inéditas
novas_frases = [
    'Gostaria de adquirir um voo para Recife',
    'Não quero mais minha reserva',
    'Preciso conversar com alguém da agência'
]

novas_frases_vec = vectorizer.transform(novas_frases)

predicoes = modelo.predict(novas_frases_vec)

# 7. Exibindo os resultados
print('--- RESULTADOS DO LAB 04 ---')

for frase, intencao in zip(novas_frases, predicoes):
    print(f"Mensagem: '{frase}' ==> Intenção Predita: [{intencao}]")
```
